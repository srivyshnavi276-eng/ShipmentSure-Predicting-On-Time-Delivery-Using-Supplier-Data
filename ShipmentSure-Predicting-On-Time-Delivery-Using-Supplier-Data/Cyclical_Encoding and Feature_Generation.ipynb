```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import PolynomialFeatures
```

```python
data = pd.read_csv("Train.csv")
```

```python
poly = PolynomialFeatures(degree=2, interaction_only=True)

p = poly.fit_transform(data[['Cost_of_the_Product','Discount_offered']])

print("Polynomial Features:")
print(poly.get_feature_names_out(['Cost_of_the_Product','Discount_offered']))

print("First 5 rows of transformed data:")
print(p[:5])
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Create hours from 0–23
hours = np.arange(24)

# Create dataframe
data2 = pd.DataFrame({'Order_Hour': hours})

# Cyclical encoding
data2['Order_Hour_sin'] = np.sin(2 * np.pi * data2['Order_Hour'] / 24)
data2['Order_Hour_cos'] = np.cos(2 * np.pi * data2['Order_Hour'] / 24)

# Plot
plt.scatter(data2['Order_Hour_sin'], data2['Order_Hour_cos'])
plt.title("Cyclical Encoding of Order Hour")
plt.axis('equal')
plt.show()
```
